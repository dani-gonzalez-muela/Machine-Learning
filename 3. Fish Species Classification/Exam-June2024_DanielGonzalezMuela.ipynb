GridSearch Parameters

```python
#Define kappa scoring function
def kappa_score(y_true, y_pred):
    return cohen_kappa_score(y_true, y_pred)

kappa_scorer = make_scorer(kappa_score)

#iterations for each Grid search
iterations=10
```

```python
components=[0]*iterations
neighbors=[0]*iterations

for i in range(iterations):    
    X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.2)
    #Cross-validation pipeline of Standarize + PCA + KNN
    pipe = Pipeline([("scale", StandardScaler()),
                     ("reduce_dims", PCA()),
                     ("knn", KNeighborsClassifier())])

    #parameters to try
    param_grid = dict(reduce_dims__n_components = [3,4,5], #decided from previous section
                      knn__n_neighbors=[1,2,3,4]) #standard values to try

    #perform cross validation in the given pipeline
    grid = GridSearchCV(pipe, param_grid=param_grid, cv=10,  scoring=kappa_scorer)
    grid.fit(X_train, y_train)
    neighbors[i]=grid.best_params_["knn__n_neighbors"]
    components[i]=grid.best_params_["reduce_dims__n_components"]

print('Parameters obtained from Grid Search:\n')
components_knn=round(np.mean(components))
neighbors_knn=round(np.mean(neighbors))
print("PCA components:", components_knn, "std(",np.std(components),")")
print("Knn neighbors:", neighbors_knn, "std(",np.std(neighbors),")")
```

**QDA**

```python
qda_components_arr=[0]*iterations
qda_regularization_arr=[0]*iterations

for i in range(iterations):    
    X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.2)
    #Cross-validation pipeline of Standarize + PCA + KNN
    pipe = Pipeline([("scale", StandardScaler()),
                     ("reduce_dims", PCA()),
                     ("qda", QDA())])

    #parameters to try
    param_grid = dict(reduce_dims__n_components = [3,4,5], #decided from previous section
                      qda__reg_param=[0, 0.05, 0.1, 0.5]) #standard values to try

    #perform cross validation in the given pipeline
    grid = GridSearchCV(pipe, param_grid=param_grid, cv=10,  scoring=kappa_scorer)
    grid.fit(X_train, y_train)
    qda_components_arr[i]=grid.best_params_["reduce_dims__n_components"]
    qda_regularization_arr[i]=grid.best_params_["qda__reg_param"]

print('Parameters obtained from Grid Search:\n')
qda_components=round(np.mean(qda_components_arr))
qda_regularization=round(np.mean(qda_regularization_arr))
print("PCA components:", qda_components, "std(",np.std(qda_components_arr),")")
print("QDA Regularization:", qda_regularization, "std(",np.std(qda_regularization_arr),")")
```

**Linear + Kernel SVM**

```python
def find_most_frequent_element(strings):
    # Dictionary to count occurrences of each element
    counts = {}
    
    # Count the occurrences of each element
    for string in strings:
        if string in counts:
            counts[string] += 1
        else:
            counts[string] = 1
    
    # Find the element with the maximum count
    most_frequent_element = max(counts, key=counts.get)
    
    return most_frequent_element
```

```python
svm_components_arr=[0]*iterations
svm_kernel_arr=[0]*iterations

for i in range(iterations):    
    X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.2)
    #Cross-validation pipeline of Standarize + PCA + KNN
    pipe = Pipeline([("scale", StandardScaler()),
                     ("reduce_dims", PCA()),
                     ("svm", SVC())])

    #parameters to try
    param_grid = dict(reduce_dims__n_components = [3,4,5], #decided from previous section
                      svm__kernel=["rbf","linear","poly"]) #standard values to try

    #perform cross validation in the given pipeline
    grid = GridSearchCV(pipe, param_grid=param_grid, cv=10,  scoring=kappa_scorer)
    grid.fit(X_train, y_train)
    svm_components_arr[i]=grid.best_params_["reduce_dims__n_components"]
    svm_kernel_arr[i]=grid.best_params_["svm__kernel"]

print('Parameters obtained from Grid Search:\n')
svm_components=round(np.mean(svm_components_arr))
svm_kernel=find_most_frequent_element(svm_kernel_arr)
print("PCA components:", svm_components, "std(",np.std(svm_components_arr),")")
print("SVM kernel:", svm_kernel, "Variability(",svm_kernel_arr,")")
```

```python
from sklearn.ensemble import RandomForestClassifier
forest_estimators_arr=[0]*iterations
forest_depth_arr=[0]*iterations

for i in range(iterations):    
    X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.2)
    #Cross-validation pipeline of Standarize + PCA + KNN
    pipe = Pipeline([("model", RandomForestClassifier())])

    #parameters to try
    param_grid = dict(model__n_estimators=[50,100],
                      model__max_depth=[10, 20,30])

    #perform cross validation in the given pipeline
    grid = GridSearchCV(pipe, param_grid=param_grid, cv=10,  scoring=kappa_scorer)
    grid.fit(X_train, y_train)
    forest_estimators_arr[i]=grid.best_params_["model__n_estimators"]
    forest_depth_arr[i]=grid.best_params_["model__max_depth"]

print('Parameters obtained from Grid Search:\n')

forest_estimators=round(np.mean(forest_estimators_arr))
forest_depth=round(np.mean(forest_depth_arr))
print("Forest Estimators:", forest_estimators, "std(",np.std(forest_estimators_arr),")")
print("Forst Depth:", forest_depth, "std(",np.std(forest_depth_arr),")")
```

**Correlated Noise:** 

```python
def generate_with_corrcoef(arr1, p):
    n = len(arr1)

    # generate noise
    noise = np.random.uniform(0, 1, n)

    # least squares linear regression for noise = m*arr1 + c
    m, c = np.linalg.lstsq(np.vstack([arr1, np.ones(n)]).T, noise,rcond=None)[0]

    # residuals have 0 correlation with arr1
    residuals = noise - (m*arr1 + c)

    # the right linear combination a*arr1 + b*residuals
    a = p * np.std(residuals)
    b = (1 - p**2)**0.5 * np.std(arr1)

    arr2 = a*arr1 + b*residuals

    # return a scaled/shifted result to have the same mean/sd as arr1
    # this doesn't change the correlation coefficient
    return np.mean(arr1) + (arr2 - np.mean(arr2)) * np.std(arr1) / np.std(arr2)

#generate correlations
noisy_features_corr=features.copy()
for i in range(sizes[-1]):
    col_name = f'C_{i+1}'
    noisy_features_corr[col_name] = generate_with_corrcoef(features[random.choice(feature_names)],i/50)
    
#train using only n correlations
sizes = [0, 10, 20, 30, 40, 50]

train_res = {x: [] for x in X }
test_res =  {x: [] for x in X}
for n in sizes:
        
    #train test
    # Fr loop for sample sizes and our models
    train_accuracy = {x: [] for x in X}
    test_accuracy = {x: [] for x in X}
    iterations=5
    
    for _ in range(iterations):
        X_train, X_test, y_train, y_test = train_test_split(noisy_features_corr.iloc[:,:n+6], labels, train_size=0.8, stratify=labels, random_state=None)

        for model, x in X.items():
            #Pipeline
            pipe = Pipeline([("scale", StandardScaler()),
                         ("model", X[model])])

            pipe.fit(X_train, y_train)
            y_train_pred = pipe.predict(X_train)
            y_test_pred = pipe.predict(X_test)

            trn_ac = cohen_kappa_score(y_train, y_train_pred)
            tst_ac = cohen_kappa_score(y_test, y_test_pred)

            train_accuracy[model].append(trn_ac)
            test_accuracy[model].append(tst_ac)       
    
    for name in X:
        train_res[name].append(np.mean(train_accuracy[name]))
        test_res[name].append(np.mean(test_accuracy[name]))
        
# Create scatter plots
plt.figure(figsize=(10, 6))

# Plot first values
plt.plot(sizes, train_res["kNN"], color='r', label='KNN Train Set',linestyle=':', marker='o')
plt.plot(sizes, test_res["kNN"], color='r', label='KNN Validation Set')

# Plot second values
plt.plot(sizes, train_res["SVM"], color='g', label='SVM Train Set',linestyle=':', marker='o')
plt.plot(sizes, test_res["SVM"], color='g', label='SVM Validation Set')

# Plot third values
plt.plot(sizes, train_res["RF"], color='b', label='RF Train Set',linestyle=':', marker='o')
plt.plot(sizes, test_res["RF"], color='b', label='RF Validation Set')

plt.xlabel('Number of Extra Noisy Features (Correlated)')
plt.ylabel('Kappa Score')
plt.title('Kappa Score vs Noise 10 Iterations')
plt.legend()
plt.grid(True)
plt.show()
```

**Subsampling: m out of n bootstrap**

```python
def m_out_of_n_bootstrap(data,labels, m, n_bootstrap):
    """
    Perform m out of n bootstrap sampling while preserving the original index.

    Parameters:
    data (pd.DataFrame): The dataset to sample from.
    m (int): The number of samples to draw in each bootstrap sample.
    n_bootstrap (int): The number of bootstrap samples to generate.

    Returns:
    List of bootstrap samples with original indices preserved.
    """
    n = len(data)
    bootstrap_samples = []
    bootstrap_labels=[]

    for _ in range(n_bootstrap):
        indices = np.random.choice(n, m, replace=True)
        sample = data.iloc[indices]
        bootstrap_samples.append(sample)
        bootstrap_labels.append(labels.iloc[indices])
        
    return bootstrap_samples,bootstrap_labels

# Example usage with n =40,000
m = 1000  # Number of samples to draw in each bootstrap sample
n_bootstrap = 10  # Number of bootstrap samples to generate SQRT OF N


bootstrap_samples,bootstrap_labels = m_out_of_n_bootstrap(features,labels, m, n_bootstrap)
```

```python
# Plotting the dendrogram for visualizing the hierarchical clustering
def plot_dendrogram(model, **kwargs):
    # Create linkage matrix and then plot the dendrogram
    counts = np.zeros(model.children_.shape[0])
    n_samples = len(model.labels_)
    for i, merge in enumerate(model.children_):
        current_count = 0
        for child_idx in merge:
            if child_idx < n_samples:
                current_count += 1  # leaf node
            else:
                current_count += counts[child_idx - n_samples]
        counts[i] = current_count

    linkage_matrix = np.column_stack([model.children_, model.distances_, counts]).astype(float)

    # Plot the corresponding dendrogram
    dendrogram(linkage_matrix, **kwargs)
```
